```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.svm import SVC, LinearSVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
import gradio as gr
```

```python
df = pd.read_csv('mnist.csv')

split = int(len(df) * 0.8)
df_train = df[:split]
df_test = df[split:]
```

```python
print(f"train: {df_train.shape}")
print(f"test: {df_test.shape}")

print(f"combined: {df.shape}")
```

```python
df.head()
```

```python
x = df.iloc[:, 1:]
y = df.iloc[:, 0]

print(f"Features: {x.shape}")
print(f"Target: {y.shape}")
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42)

print(f"Training data: {x_train.shape[0]}")
print(f"Test data: {x_test.shape[0]}")
```

```python
model = SVC(kernel='rbf', C=1)

model.fit(x_train, y_train)
```

```python
from sklearn.metrics import classification_report
from sklearn.model_selection import cross_val_score


predictions = model.predict(x_test)

accuracy = accuracy_score(y_test, predictions)

print(f"Accuracy: {accuracy:.4f} ({accuracy * 100:.2f}%)")

print(classification_report(y_test, predictions))

# scores = cross_val_score(model, x, y, cv=5)
# print(scores)
# print(f"Medel: {scores.mean():.4f}, Std: {scores.std():.4f}")
```

```python
import json
import numpy as np

model_data = {
    "model_type": "SVC",
    "kernel": "rbf",
    "C": 1,
    "support_vectors": model.support_vectors_.tolist(),
    "dual_coef": model.dual_coef_.tolist(),
    "intercept": model.intercept_.tolist(),
    "classes": model.classes_.tolist(),
    "n_support": model.n_support_.tolist(),
    "accuracy": 0.9635
}

with open("svm_model.json", "w") as f:
    json.dump(model_data, f, indent=2)

print("Modellen sparad som svm_model.json")
```
